# MERFISH three-stage reconstruction
Spatial point set -> predicted cell type -> GiT conditional expression.

Default: one fold, three training stages. Shuffles add generation/evaluation only. Existing RC outputs are separate.


```python
from pathlib import Path
import json, sys

candidates = [Path.cwd(), Path.cwd() / 'ST_DeepSpatial', Path.cwd().parent]
PROJECT = next(p.resolve() for p in candidates if (p / 'stspatial').is_dir())
sys.path.insert(0, str(PROJECT))
from stspatial.experiment import run
```

```python
CONFIG = json.loads((PROJECT / 'configs/merfish.json').read_text(encoding='utf-8-sig'))
OUTPUT = PROJECT / 'outputs/merfish_staged_v1'
HOLDOUTS = [2]  # Use [1, 2, 3] for three folds.
SEEDS = [42]
VARIANTS = ['full']
INTERVENTIONS = True
STAGE = 'all'
DEVICE = 'auto'
RESUME = True
EXCLUDED_OUTER_TEST = []
ROLE = 'development'
print('Output:', OUTPUT)
print('Full-model training stages:', 3 * len(HOLDOUTS) * len(SEEDS))
```

```python
records = run(CONFIG, OUTPUT, HOLDOUTS, SEEDS, VARIANTS, stage=STAGE,
              device=DEVICE, resume=RESUME, excluded=EXCLUDED_OUTER_TEST,
              role=ROLE, interventions=INTERVENTIONS)
print('Completed evaluation records:', len(records))
print('Saved to:', OUTPUT.resolve())
```
